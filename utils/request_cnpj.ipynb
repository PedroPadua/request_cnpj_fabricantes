```python
import requests

base_url = "https://publica.cnpj.ws/cnpj"
cnpj = "47986104000180"

url = f"{base_url}/{cnpj}"
busca = ['Fabricação','Confecção','Produção','Preparação','Beneficiamento','Processamento','Torrefação','Moagem','Refino','Fundição','Forjamento','Usinagem','Montagem','Impressão','Reciclagem','Transformação','Elaboração','Manipulação']
response = requests.get(url)

if response.status_code == 200:

    data = response.json()
    estabelecimento = data.get('estabelecimento')
    ati_prin = estabelecimento['atividade_principal']['descricao']

    for termo in busca:
        if termo in ati_prin:
            print(f'Contém {termo} na atividade principal: {ati_prin}')
            break 
        
    if len(estabelecimento['atividades_secundarias']) > 0:
        for ati_sec in estabelecimento.get('atividades_secundarias', []):
            desc_sec = ati_sec['descricao']
            for termo in busca:
                if termo in desc_sec:
                    print(f'Contém {termo} na secundária: {desc_sec}')
                    break
else:
    print(f"Erro {response.status_code}: {response.json()}")
```
